```python

from matplotlib import pyplot as plt

img = cv2.imread('image3.jpeg')

w = 500
h = 500
dimensions = (w, h)

resize_img = cv2.resize(img, dimensions, interpolation=cv2.INTER_LINEAR)

cv2.imwrite('image3_resized.jpeg', resize_img)

resize_img_rgb = cv2.cvtColor(resize_img, cv2.COLOR_BGR2RGB)

plt.imshow(resize_img_rgb)
plt.title('Resized Image')
plt.axis('off')
plt.show()
```

```python
#Rotate the image with 120 degrees.
img = cv2.imread('image3.jpeg')

(h, w) = img.shape[:2]
center = (w // 2, h // 2)


M = cv2.getRotationMatrix2D(center, 120, 1.0)
rotated = cv2.warpAffine(img, M, (w, h))

cv2.imwrite('rotated_120.jpeg', rotated)
plt.imshow(cv2.cvtColor(rotated, cv2.COLOR_BGR2RGB))
plt.title('Rotated')
plt.axis('off')
plt.show()
```

```python
# Shear matrix
shear_matrix = np.float32([
    [1, 0.5, 0],   #
    [0, 1, 0]
])

sheared = cv2.warpAffine(img, shear_matrix, (w, h))

cv2.imwrite('sheared.jpeg', sheared)
plt.imshow(cv2.cvtColor(sheared, cv2.COLOR_BGR2RGB))
plt.title('Sheared')
plt.axis('off')
plt.show()
```
